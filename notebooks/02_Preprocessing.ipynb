# Data Preprocessing for JIT Bug Prediction

## Objective

The purpose of this notebook is to prepare the ApacheJIT dataset
for machine learning experiments.

The preprocessing pipeline includes:

- Data loading
- Target variable conversion
- Chronological ordering
- Feature selection
- Missing value handling
- Saving a clean processed dataset

```python
import pandas as pd
import numpy as np
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
path = "/content/drive/MyDrive/JIT_bug_prediction/data/apachejit.csv"

df = pd.read_csv(path)

df.head()
```

```python
print("Dataset Shape:", df.shape)
```

## Target Variable Conversion

The target variable `buggy` is converted into binary format.

0 → Non-Buggy Commit

1 → Buggy Commit

```python
df['buggy'] = (
    df['buggy']
    .astype(str)
    .str.lower()
    .map({
        'false':0,
        'true':1
    })
)
```

```python
df['buggy'].value_counts()
```

## Chronological Ordering

The dataset is sorted using author_date.

This preserves temporal order and enables
realistic future commit prediction.

```python
df = df.sort_values(
    by='author_date'
)

df.head()
```

## Feature Selection

The selected features are commonly used
JIT software metrics.

```python
selected_columns = [

    'author_date',

    'la',
    'ld',
    'nf',
    'nd',
    'ns',
    'ent',
    'ndev',
    'age',
    'nuc',
    'aexp',
    'arexp',
    'asexp',

    'buggy'
]
```

```python
processed_df = df[selected_columns]
```

```python
processed_df = processed_df.fillna(0)
```

```python
processed_df.isnull().sum()
```

```python
processed_df.head()
```

## Saving Processed Dataset

The cleaned dataset is saved for use
in all future machine learning experiments.

```python
processed_path = "/content/drive/MyDrive/JIT_bug_prediction/data/processed_data.csv"

processed_df.to_csv(
    processed_path,
    index=False
)

print("Saved Successfully")
```
